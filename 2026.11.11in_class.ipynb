```python
import pandas as pds
import numpy as np
import matplotlib.pyplot as plt
import seaborn
```

```python
y = np.array([3,3,3,4,4,12,12,13])
len_y = len(y)
```

```python
seaborn.lineplot(x=range(len_y), y=y)
plt.show()
```

```python
y + np.random.normal(loc=0, scale=6, size=len_y)
```

```python
seaborn.lineplot(x=range(len_y), y=y + np.random.normal(loc=0, scale=6, size=len_y), lw=100, color = 'chartreuse')
plt.show()
```

```python
seaborn.lineplot(x=range(1, len(sample)+1), y=sample, lw=30, color = 'black')
seaborn.lineplot(x=range(1, len(sample2)+1), y=sample2, lw=30, color='black', alpha=0.7)
seaborn.lineplot(x=range(1, len(sample3)+1), y=sample3, lw=30, color='black', alpha=0.3)
plt.xlabel('Number of excised introns')
plt.ylabel('New introns excised')
plt.show()
```
